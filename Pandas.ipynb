## 15. Crie uma nova coluna "Faixa de Preço" usando a coluna "Preço Unitário" para classificar os produtos como:

## Baixo (preço < 50)
## Médio (50 ≤ preço ≤ 300)
## Alto (preço > 300)

```python
df.loc[df['Preço Unitário'] < 50, 'Faixa de Preço'] = 'Baixo'
df.loc[(df['Preço Unitário'] >= 50) & (df['Preço Unitário'] <= 300), 'Faixa de Preço'] = 'Médio'
df.loc[df['Preço Unitário'] > 300, 'Faixa de Preço'] = 'Alto'
display(df)
```

```python
df["Faixa de Preço"].value_counts().reset_index()
```

##16.Crie uma nova coluna "Status Simplificado" para transformar a coluna "Status do Pedido":
##"Entregue" → "Finalizado"
##"Em andamento" → "Processando"
##"Cancelado" → "Cancelado"

```python
df.loc[df['Status do Pedido'] == 'Entregue', 'Status Simplificado'] = 'Finalizado'
df.loc[df['Status do Pedido'] == 'Em andamento', 'Status Simplificado'] = 'Processando'
df.loc[df['Status do Pedido'] == 'Cancelado', 'Status Simplificado'] = 'Cancelado'
df.loc[df['Status do Pedido'].isnull(), 'Status Simplificado'] = 'Desconhecido'
df.head()
```

## 17. Crie uma nova coluna "Preço com Desconto", aplicando 10% de desconto sobre o valor original da venda.

```python
df['Preço com Desconto'] = round(df['Valor da Venda'] - (df['Valor da Venda'] * 10 / 100), 2)
df.head()
```

## 18. Quantos produtos estão na faixa de preço "Alto"?

```python
df[df["Faixa de Preço"] == "Alto"]["Faixa de Preço"].value_counts().reset_index()
```
