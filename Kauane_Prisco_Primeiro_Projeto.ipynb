Será necessário converte o DataFrame em listas, onde cada coluna será uma lista.
Será necessario realizar uma limpeza dos dados, removendo ou corrigindo valores inválidos, como:
Remover valores NaN ou strings vazias.
Garantir que os tipos de dados estejam corretos:
O campo salário deve ser convertido para o tipo float.
O campo idade deve ser convertido para o tipo int.
O sistema deve autenticar o usuário no Google Cloud usando a biblioteca google-cloud-storage.
O sistema deve realizar o upload do arquivo tratado para um bucket no Google Cloud Storage.
O arquivo tratado deve ser salvo no formato CSV dentro do bucket com o nome apropriado

```python
import pandas as pd
```

```python
df = pd.read_csv('https://storage.googleapis.com/projeto-kauane-prisco-1/dados-brutos/tabela_funcionarios.csv')
```

```python
display(df)
```

```python
nome = list(df.nome)
idade = list(df.idade)
salario = list(df.salario)
cargo = list(df.cargo)
print(nome)
print(idade)
print(salario)
print(cargo)
```

```python
dados_tratados = []

for x, y, z, n in zip(nome, idade, salario, cargo):
    dados_tratados.append([x, y, z, n])

for index,linha in enumerate(dados_tratados):
  print(index,linha)
```

```python
for index, linha in enumerate(dados_tratados):
    if any(elemento != elemento for elemento in linha):  # Verifica se há NaN na linha
        print(f'\n{index}: {linha}')
```

```python
dados_tratados.pop(14)
dados_tratados.pop(12)
dados_tratados.pop(7)
dados_tratados.pop(4)
dados_tratados.pop(3)
dados_tratados.pop(2)
```

```python
for index,linha in enumerate(dados_tratados):
  print(index,linha)
```

```python
#Criando listas com nome, idade, salario e cargo.
'''
Garantir que os tipos de dados estejam corretos:
O campo salário deve ser convertido para o tipo float.
O campo idade deve ser convertido para o tipo int.
'''
nome = []
idade = []
salario = []
cargo = []

for linha in dados_tratados:
    nome.append(linha[0])
    idade.append(int(float(linha[1])))
    salario.append(float(linha[2]))
    cargo.append(linha[3])

# Exibindo as listas
print("nome:", nome)
print(type(nome[0]))
print("idade:", idade)
print(type(idade[0]))
print("salario:", salario)
print(type(salario[0]))
print("cargo:", cargo)
print(type(cargo[0]))
```

```python
dados_tratados_novo = [["nome", "idade", "salario", "cargo"]]
for i in range(len(nome)):
    dados_tratados_novo.append([nome[i], idade[i], salario[i], cargo[i]])
#print(dados_tratados_novo)

for index,linha in enumerate(dados_tratados_novo):
  print(index,linha)
```

```python
df = pd.DataFrame({
    'nome': nome,
    'idade': idade,
    'salario': salario,
    'cargo': cargo
})
```
