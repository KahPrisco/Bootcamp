##Crie uma nova coluna chamada Nota_Final, que seja a média entre Nota_Prova1, Nota_Prova2 e Nota_Trabalho.

```
df_escola["Nota_Final"] = df_escola[["nota_prova1", "nota_prova2", "nota_trabalho"]].mean(axis=1).round(1)

df_escola
```

## Preencha os valores nulos com a média do aluno

```
notas = ["nota_prova1", "nota_prova2", "nota_trabalho"]

media_alunos = df_escola[notas].mean(axis=1).round(1)
df_escola[notas] = df_escola[notas].T.fillna(media_alunos).T

media_geral_horas = df_escola['horas_estudo'].mean().round(1)
df_escola['horas_estudo'] = df_escola['horas_estudo'].fillna(media_geral_horas)

display(df_escola)

```

##Crie uma coluna Situação com base na Nota_Final:

Aprovado: nota final ≥ 7

Recuperação: 5 ≤ nota final < 7

Reprovado: nota final < 5

```
df_escola['Situação'] = 'Reprovado'
df_escola.loc[(df_escola['Nota_Final'])>=7, 'Situação'] = 'Aprovado'
df_escola.loc[(df_escola['Nota_Final']>=5) & (df_escola['Nota_Final']<7),'Situação' ] = 'Recuperação'

display(df_escola)
```

Calcule a **mediana** das horas de estudo.


```
mediana_horas_estudos = np.median(df_escola.horas_estudo	)
print(f'A mediana das horas de estudos é:{mediana_horas_estudos}')
```

Calcule a **moda** das notas finais.

```
moda_notas_finais = st.mode(df_escola.Nota_Final)
print(f'A moda das notas finais é:{moda_notas_finais}')
```

 Calcule a **variância** das idades.

```
variancia_idade = st.variance(df_escola.idade)
print(f'A variância das idades é:{variancia_idade}')
```

Calcule o **desvio padrão** das horas de estudo.

```
desvio_padrao_horas_estudos = st.stdev(df_escola.horas_estudo)
print(f'O desvio padrão das horas de estudo é:{desvio_padrao_horas_estudos}')
```

Calcule a **amplitude interquartílica (AIQ)** das notas finais.

```
q1st = st.quantiles(df_escola.Nota_Final)[0]
print(f'Q1 (notas finais) = {q1st}')


q2st = st.quantiles(df_escola.Nota_Final)[1]
print(f'Q2 (notas finais) = {q2st}')


q3st = st.quantiles(df_escola.Nota_Final)[2]
print(f'Q3 (notas finais) = {q3st}')

aiqst = q3st - q1st
print("\nAmplitude Interquatilica (AIQ):", aiqst)
```

Crie a **tabela de frequência absolutada coluna Nota_Final**.

Adicione a **frequência acumulada**.

Identifique a **nota com maior frequência**.

```
# Criando uma tabela de frequências para variável quantitativa

faixas = ['[00 - 02[','[02 - 04[','[04 - 06[','[06 - 08[','[08 - 10]']
notas_em_faixas = pd.cut(x=df_escola['Nota_Final'], bins=[0, 1, 3, 5, 7, 10], labels=faixas)


# Calculando a frequência absoluta e percentual
freq = notas_em_faixas.value_counts()
percentual = notas_em_faixas.value_counts(normalize=True) * 100

# Criando o DataFrame com frequências
Notas_Finais = pd.DataFrame({'freq_abs': freq, 'freq_%': percentual})

# Ordenando pelo índice para garantir a ordem das faixas
Notas_Finais = Notas_Finais.sort_index()

# Resetando o índice
Notas_Finais = Notas_Finais.reset_index()
Notas_Finais.rename(columns={'index': 'faixa'}, inplace=True)

# Calculando a frequência acumulada
Notas_Finais['freq_acum'] = Notas_Finais['freq_abs'].cumsum()
Notas_Finais['freq_acum_%'] = (Notas_Finais['freq_acum'] / Notas_Finais['freq_abs'].sum()) * 100

# Adicionando a linha "Total"
Notas_Finais.loc[len(Notas_Finais)] = ['Total', Notas_Finais['freq_abs'].sum(), Notas_Finais['freq_%'].sum(), '-', '-']

# Arredondando os valores percentuais
Notas_Finais['freq_%'] = Notas_Finais['freq_%'].round()
Notas_Finais['freq_acum_%'] = Notas_Finais['freq_acum_%'].round()

# Exibindo a tabela final
display(Notas_Finais)


print("\nA nota com maior frequência absoluta esta na faixa [08 - 10]")
```

 Interprete: o que significa um valor estar entre Q1 e Q3 ?

Q1 representa o percentil de 25% e Q3 representa o percentil de 75%.
Portanto, estar entre Q1 e Q3 significa estar dentro do intervalo entre o 25º e o 75º percentil.
Isso corresponde a 75% - 25% = 50%, ou seja, o valor está situado dentro dos 50% centrais do conjunto de dados, também conhecido como intervalo interquartil (IQR).
